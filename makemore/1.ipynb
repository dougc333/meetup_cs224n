```python
fd = open('names.txt','r')
data = fd.readlines()
arr_names = [x.strip() for x in data]
print(len(arr_names),arr_names[:10])
```

```python
for word in arr_names[:1]:
    for ch1, ch2 in zip(word,word[1:]):
        print (ch1, ch2)
```


$Pr(bigram(em)) = \frac{Count(bigram(em)) \space in \space corpus}{Count(bigrams \space start \space with \space letter \space e)}$

```python
#add start and stop tokens to distinguish start and end of words. Space is ambiguous
for word in arr_names[:1]:
    chs = ['<S>'] + list(word)+['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        print (ch1, ch2)
```
