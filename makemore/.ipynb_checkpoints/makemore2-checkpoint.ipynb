```python
#https://www.youtube.com/watch?v=5TMKYF4o4iE


class PCA:
    def __init__(self, n_components=2):
        self.n_components = n_components
    def fit(self,X):
        self.mean = np.mean(X, axis=0)
        X = X-self.mean
        print("centered X:",X)
        cov = np.cov(X.T)
        print("cov:",cov)
        eigVal, eigVec = np.linalg.eigh(cov)
        print("eigVal shape:",eigVal.shape)
        print("eigVec shape:",eigVec.shape)
        print("eigVal",eigVal)
        print("eigVec",eigVec)
        
        idxs = np.argsort(eigVal)[::-1]
        eigVal = eigVal[idxs]
        eigVec = eigVec[:,idxs]
        self.n_components = eigVec[:,0:self.n_components]
    def transform(self,X):
        X = X-self.mean
        return np.dot(X, self.n_components)

np.random.seed(1)
X = np.random.rand(3, 10)
print("X:",X)
pca = PCA(2)
pca.fit(X)
X_pca = pca.transform(X)
print("final:",X_pca)
```

```python
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
print(itos)
```

```python
block_size = 3
X, Y = [], []
for w in words[:3]:
    print(w)
    context = [0] * block_size
    for ch in w + '.':
        ix = stoi[ch]
        X.append(context)
        Y.append(ix)
        print(''.join(itos[i] for i in context),'--->', itos[ix])
        context = context[1:] + [ix]
X = torch.tensor(X)
Y = torch.tensor(Y)
```
