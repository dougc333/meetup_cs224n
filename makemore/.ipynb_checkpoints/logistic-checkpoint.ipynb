# sentiment analysis
<h4> logistic regression applies a sigmoid to the output of linear regression to obtain a 0/1 output</h4>

<h6>Start with a linear regression equation</h6>
$y_n = \sum_0^N w_i * x_i + b$

<h6>A sigmoid defined below is a continuous function converting a linear range into a 0 or 1. There is a transition between 0 and 1 which looks like a vertical line but is differentiable</h6>
sigmoid(z) = $\frac{1}{1+exp(-z)} = \frac{e^z}{1+e*z}$
